```python
import panel as pn
import xarray as xr
import hvplot.xarray
import pandas as pd
import panel.widgets as pnw
```

```python
pn.extension()
```

```python
ds = xr.tutorial.load_dataset('air_temperature')
ds
```

```python
time_values = ds.time.to_numpy()
time_values
```

```python
datetime_range_slider = pn.widgets.DatetimeRangeSlider(
    start=time_values[0],
    end=time_values[-1],
    name='Dates',
    step=21600000
)
```

```python
def select_range(date_range):
    start = pd.Timestamp(date_range[0])
    end = pd.Timestamp(date_range[1])
    return ds.air[(ds.time >= start) & (ds.time <= end)]

data_frame = hvplot.bind(select_range, datetime_range_slider).interactive()
```

```python
data_frame.sel(lat=pnw.DiscreteSlider, lon=pnw.DiscreteSlider).hvplot()
```
